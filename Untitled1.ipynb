```python
import pandas as pd
import numpy as np
import os
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Define file paths
base_path = r"C:\Users\Admin\Desktop\Mac files\Signal Processing\CMaps"

train_files = ["train_FD001.txt", "train_FD002.txt", "train_FD003.txt", "train_FD004.txt"]
test_files = ["test_FD001.txt", "test_FD002.txt", "test_FD003.txt", "test_FD004.txt"]
rul_files = ["RUL_FD001.txt", "RUL_FD002.txt", "RUL_FD003.txt", "RUL_FD004.txt"]

# Load the datasets
train_data = {f"FD00{i}": pd.read_csv(os.path.join(base_path, train_files[i-1]), sep='\s+', header=None) for i in range(1, 5)}
test_data = {f"FD00{i}": pd.read_csv(os.path.join(base_path, test_files[i-1]), sep='\s+', header=None) for i in range(1, 5)}
rul_data = {f"FD00{i}": pd.read_csv(os.path.join(base_path, rul_files[i-1]), sep='\s+', header=None) for i in range(1, 5)}

# Assign column names to datasets
columns = ['engine_id', 'cycle', 'operational_setting_1', 'operational_setting_2', 'operational_setting_3'] + \
          [f'sensor_{i}' for i in range(1, 22)]

for key in train_data.keys():
    train_data[key].columns = columns
    test_data[key].columns = columns
```

```python
def preprocess_data(train_data, test_data, rul_data, key, n_components=5):
    constant_columns = ['operational_setting_3', 'sensor_1', 'sensor_5', 'sensor_16', 'sensor_18', 'sensor_19']
    train_data[key] = train_data[key].drop(columns=constant_columns)
    test_data[key] = test_data[key].drop(columns=constant_columns)

    # Ensure numerical columns are float64 before normalization
    numerical_columns = train_data[key].select_dtypes(include=['int64', 'float64']).columns[2:]
    train_data[key][numerical_columns] = train_data[key][numerical_columns].astype('float64')
    test_data[key][numerical_columns] = test_data[key][numerical_columns].astype('float64')

    # Normalize features
    scaler = MinMaxScaler()
    normalized_train = scaler.fit_transform(train_data[key].iloc[:, 2:].values)
    normalized_test = scaler.transform(test_data[key].iloc[:, 2:].values)

    train_data[key].iloc[:, 2:] = normalized_train
    test_data[key].iloc[:, 2:] = normalized_test

    # Apply PCA for feature reduction
    pca = PCA(n_components=n_components)
    feature_columns = [col for col in train_data[key].columns if 'sensor_' in col]
    train_pca = pca.fit_transform(train_data[key][feature_columns])
    test_pca = pca.transform(test_data[key][feature_columns])

    # Replace sensor columns with PCA components
    for i in range(n_components):
        train_data[key][f'pca_{i+1}'] = train_pca[:, i]
        test_data[key][f'pca_{i+1}'] = test_pca[:, i]
    train_data[key] = train_data[key].drop(columns=feature_columns)
    test_data[key] = test_data[key].drop(columns=feature_columns)

    # Add RUL for training data
    train_data[key]['RUL'] = train_data[key].groupby('engine_id')['cycle'].transform('max') - train_data[key]['cycle']

    # Map RUL values to test data
    rul_mapping = {i+1: rul_data[key].iloc[i, 0] for i in range(rul_data[key].shape[0])}
    test_data[key]['RUL'] = test_data[key]['engine_id'].map(rul_mapping) + \
                            test_data[key].groupby('engine_id')['cycle'].transform('max') - test_data[key]['cycle']

    print(f"--- Debugging {key} ---")
    print(f"RUL Mapping for {key} (First 5):")
    print(test_data[key][['engine_id', 'cycle', 'RUL']].head())
    return train_data, test_data

# Preprocess all datasets
for key in train_data.keys():
    train_data, test_data = preprocess_data(train_data, test_data, rul_data, key)
```

```python
# Verification function to check preprocessed data
import numpy as np

def verify_preprocessed_data(train_data, test_data, rul_data, original_train_files, original_test_files, original_rul_files):
    for key, original_train_file, original_test_file, original_rul_file in zip(
        train_data.keys(), original_train_files, original_test_files, original_rul_files
    ):
        print(f"\n--- Verifying Preprocessed Data for {key} ---")

        # Load raw data
        raw_train = pd.read_csv(original_train_file, sep='\s+', header=None)
        raw_test = pd.read_csv(original_test_file, sep='\s+', header=None)
        raw_rul = pd.read_csv(original_rul_file, sep='\s+', header=None)

        # Check dimensions
        if len(train_data[key]) != len(raw_train):
            print(f"Mismatch in train data length for {key}.")
            return
        if len(test_data[key]) != len(raw_test):
            print(f"Mismatch in test data length for {key}.")
            return

        # Verify RUL
        mapped_rul = train_data[key]['RUL']
        computed_rul = (
            train_data[key].groupby('engine_id')['cycle'].transform('max') - train_data[key]['cycle']
        )
        if not np.allclose(mapped_rul, computed_rul):
            print(f"Mismatch in RUL values for training data in {key}.")
            return

        mapped_rul_test = test_data[key]['RUL']
        rul_mapping = {i + 1: raw_rul.iloc[i, 0] for i in range(raw_rul.shape[0])}
        computed_rul_test = test_data[key]['engine_id'].map(rul_mapping) + \
            test_data[key].groupby('engine_id')['cycle'].transform('max') - test_data[key]['cycle']
        if not np.allclose(mapped_rul_test, computed_rul_test):
            print(f"Mismatch in RUL values for test data in {key}.")
            return

        # Print if all checks pass
        print(f"All checks passed for {key}.")

# Define file paths for verification
original_train_files = [
    os.path.join(base_path, file) for file in train_files
]
original_test_files = [
    os.path.join(base_path, file) for file in test_files
]
original_rul_files = [
    os.path.join(base_path, file) for file in rul_files
]

# Run verification
verify_preprocessed_data(train_data, test_data, rul_data, original_train_files, original_test_files, original_rul_files)
```

```python
class RULDataset(Dataset):
    def __init__(self, dataframe):
        self.data = dataframe

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        row = self.data.iloc[idx]
        features = torch.tensor(row.iloc[2:-1].values, dtype=torch.float32)
        label = torch.tensor(row.iloc[-1], dtype=torch.float32)
        return features, label

# Prepare DataLoaders
batch_size = 64
train_loaders = {}
test_loaders = {}

for key in train_data.keys():
    train_dataset = RULDataset(train_data[key])
    test_dataset = RULDataset(test_data[key])

    train_loaders[key] = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_loaders[key] = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
class GNNRULPredictor(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(GNNRULPredictor, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)
        self.fc3 = nn.Linear(hidden_dim, output_dim)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
# Training loop with early stopping
def train_with_early_stopping(model, dataloader, optimizer, criterion, patience=5):
    best_loss = float("inf")
    patience_counter = 0

    for epoch in range(50):  # Maximum 50 epochs
        total_loss = 0.0
        model.train()
        for features, labels in dataloader:
            optimizer.zero_grad()
            predictions = model(features)
            loss = criterion(predictions.view(-1), labels)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        avg_loss = total_loss / len(dataloader)
        print(f"Epoch {epoch+1}, Loss: {avg_loss:.4f}")

        if avg_loss < best_loss:
            best_loss = avg_loss
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break
    return model

# Testing loop
def test_model(model, dataloader):
    model.eval()
    predictions, actuals = [], []
    with torch.no_grad():
        for features, labels in dataloader:
            pred = model(features)
            predictions.extend(pred.view(-1).tolist())
            actuals.extend(labels.tolist())
    return predictions, actuals

# Main training and evaluation process
results = {}
input_dim = 7  # Based on preprocessed feature size
hidden_dim = 128
output_dim = 1

for key in train_loaders.keys():
    print(f"Training for {key}...")

    # Initialize the model
    model = GNNRULPredictor(input_dim=input_dim, hidden_dim=hidden_dim, output_dim=output_dim)
    optimizer = optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.MSELoss()

    # Train the model with early stopping
    model = train_with_early_stopping(model, train_loaders[key], optimizer, criterion, patience=5)

    # Test the model
    predictions, actuals = test_model(model, test_loaders[key])

    # Evaluate performance
    rmse = np.sqrt(mean_squared_error(actuals, predictions))
    mape = mean_absolute_percentage_error(actuals, predictions)
    accuracy = 100 - mape * 100

    print(f"{key} - RMSE: {rmse:.2f}, MAPE: {mape:.2%}, Accuracy: {accuracy:.2f}%")
    results[key] = (predictions, actuals)

# Visualization of results
import matplotlib.pyplot as plt

for key, (predictions, actuals) in results.items():
    plt.figure(figsize=(10, 6))
    plt.scatter(actuals, predictions, alpha=0.5, label=f'{key} Predictions')
    plt.plot([min(actuals), max(actuals)], [min(actuals), max(actuals)], 'r--', label='Ideal Fit')
    plt.xlabel('Actual RUL')
    plt.ylabel('Predicted RUL')
    plt.title(f'Actual vs Predicted RUL for {key}')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def visualize_results_with_metrics(results):
    for key, (predictions, actuals) in results.items():
        rmse = np.sqrt(mean_squared_error(actuals, predictions))
        mape = mean_absolute_percentage_error(actuals, predictions)
        accuracy = 100 - mape * 100

        plt.figure(figsize=(10, 6))
        plt.scatter(actuals, predictions, alpha=0.6, label="Predictions")
        plt.plot([min(actuals), max(actuals)], [min(actuals), max(actuals)], color="red", linestyle="--", label="Ideal")
        plt.title(f"Actual vs Predicted RUL ({key})\nRMSE: {rmse:.2f}, MAPE: {mape:.2%}, Accuracy: {accuracy:.2f}%")
        plt.xlabel("Actual RUL")
        plt.ylabel("Predicted RUL")
        plt.legend()
        plt.grid(True)
        plt.show()

visualize_results_with_metrics(results)
```
